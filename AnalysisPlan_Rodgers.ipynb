My hypothesis is that food inflation rates are higher during the holiday months (October-December). This is my hypothesis because people often gather together during the holiday season to cook and have meals together. 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
rates = pd.read_csv(r'/Users/clarencerodgers/Downloads/FoodInflationRates_Revised.csv')
print(rates)
```

To help me with my analysis, I took the data from the US Inflation Calculator website and made a new Google Sheets with a Year, Month, and Rates column. I did this so that I could more easily create charts that would compare the food inflation rates of each month side by side. 

Instead of looking at all the data from 1968-2024, I decided to just stick with a 10 year time span (2014-2024) so the chart would not be too long. 

```python
sns.boxplot(x="Month", y="Rates", data=rates)
plt.title("Food Price Percent Increases per Month (2014-2024)")
```

I decided to use boxplots to display this dataset because it allows me to visually see the distribution of the food inflation rates of each month side by side. With the boxplots, I can see the difference in medians. If the boxes are higher up, that means that the median and the values within the interquartile range are higher, meaning that the month likely has higher food inflation rates. I also decided to use boxplots because it allows me to see the highest value of inflation rate for each month. 

From looking at the boxplots, the data seems to be telling me that the food inflation rates are roughly evenly distributed across the months. There is no month with an interquartile range and median that have noticably higher rates than the others. November and December seem to have the highest medians. October and December are tied for the 4th largest Q3 value, behind August (largest Q3 value), September (2nd largest Q3 value), and July (3rd largest Q3 value) The month of August has the largest outlier. 

My analysis of this data is that while November and December having the highest medians helps support my hypothesis of the holiday months having the highest food inflation rates, I cannot confirm that my hypothesis is correct because there are others months that have higher maximum values, and higher Q3 values (such as August). 

**To help me better answer my hypothesis, I am going to create a bar graph displaying the mean food inflation price percent increase for each month side by side**

To do this, I am first going to have Python calculate the mean price percent increase of each month:

```python
avg = rates.groupby("Month")["Rates"].mean()
avg.head()
```

I am then going to use this to create a bar graph:

```python
avg.plot(kind="bar", figsize=(8,4), color="green")
plt.title("Average Percent Inflation Increase per Month (from 2014-2024)")
plt.xlabel("Month")
plt.ylabel("Average Percent Price Increase")
plt.show()
```

I decided to use a bar graph because bar graphs allow for a good visual comparison of values. It helps to clearly see which month has the highest mean price percent increase for food inflation. 

From the bar graph, we can see that the mean price percent increase for food inflation is roughly the same across all the months. December has the highest mean out of all the months, with November's mean being the second highest, followed by September. (I held a ruler up to my screen to help since the bars are pretty similar in size)

Since December and November have the 2 highest mean percent increases for food inflation rates, this supports my hypothesis that food inflation rates are higher during the holiday months. However, I do not think I can outright confirm my hypothesis since all the bars in the bar graph are so close in size.
