```python
x = months["Rates"].to_numpy()
y = months["Year"].to_numpy()

spearman_rho, spearman_p = stats.spearmanr(x,y)
print(f"{spearman_rho:.3f}")
```

## Data Wrangling

### Boxplot

In order to test the hypothesis that food inflation rates are higher during the holiday months (October-December), I am going to make a multivariate boxplot to compare the food inflation rates of each month over the past ten years side by side. This way, I can see the median inflation rate of each month, as well as the maximum inflation rate. Seeing this can help me to test my hypothesis because the month with the highest median and maximum rate likely has the higher inflation rates overall.

I will be using the "Month" column and the "Rates" column to help me test the hypothesis, since I want to find which month has the highest inflation rates.

```python
sns.boxplot(x="Month", y="Rates", data=months)
plt.title("Food Price Percent Increases per Month (2014-2024)")
```

From looking at the boxplots, the data seems to be telling me that the food inflation rates are roughly evenly distributed across the months. There is no month with an interquartile range and median that have noticably higher rates than the others. November and December seem to have the highest medians. October and December are tied for the 4th largest Q3 value, behind August (largest Q3 value), September (2nd largest Q3 value), and July (3rd largest Q3 value) The month of August has the largest outlier.

### Bar Graph

I am also going to make a bar graph comparing the average price percent increases of each month to help me test my hypothesis as well. I am comparing the mean price percent increase of food prices each month because it will help me see which month, on average, has the highest food inflation rates, which will help test if the inflation rates are highest during the holiday months. Looking at two visualizations and comparing more than one type of statistic will help to better answer the hypothesis.  

I am using a bar graph because because bar graphs allow for a good visual comparison of values. It helps to clearly see which month has the highest mean price percent increase for food inflation.

To make the bar graph, I am going to use Python to find the mean price percent increase of each month.

```python
avg = months.groupby("Month")["Rates"].mean()
avg.head()
```

I am then going to use this to create a bar graph:

```python
avg.plot(kind="bar", figsize=(8,4), color="green")
plt.title("Average Percent Inflation Increase per Month (from 2014-2024)")
plt.xlabel("Month")
plt.ylabel("Average Percent Price Increase")
plt.show()
```

From the bar graph, we can see that the mean price percent increase for food inflation is roughly the same across all the months. December has the highest mean out of all the months, with November's mean being the second highest, followed by September. (I held a ruler up to my screen to help since the bars are pretty similar in size)
